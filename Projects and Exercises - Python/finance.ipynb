```python
import ta

def add_indicators(group):
    # group = group.sort_values('Date')  # IMPORTANT
    group['rsi'] = ta.momentum.RSIIndicator(close=group['Close']).rsi()
    group['sma_20'] = ta.trend.SMAIndicator(close=group['Close'], window=20).sma_indicator()
    group['sma_50'] = ta.trend.SMAIndicator(close=group['Close'], window=50).sma_indicator()
    group['sma_100'] = ta.trend.SMAIndicator(close=group['Close'], window=100).sma_indicator()
    group['sma_200'] = ta.trend.SMAIndicator(close=group['Close'], window=200).sma_indicator()
    group['macd'] = ta.trend.macd(close=group['Close'])
    group['atr'] = ta.volatility.AverageTrueRange(close=group['Close'], high=group['High'],low=group['Low'], window=10).average_true_range()
    group['ease_of_move'] = ta.volume.EaseOfMovementIndicator(high=group['High'], low=group['Low'], volume=group['Volume'], window=14).ease_of_movement()
    
    # look up additional indicators
    
    # Add price to MA ratios as additional features
    
    
    return group


df3 = df2.groupby('symbol').filter(lambda x: len(x) > 1000)
df3 = df3.groupby('symbol').apply(add_indicators)
```

```python
def signal(row):
    if (row['rsi'] > (df2["rsi"].mean() - df2["rsi"].std())) and (row['rsi'] < (df2["rsi"].mean() + df2["rsi"].std())):
        if (row['macd'] > (df2["macd"].mean() - df2["macd"].std())) and (row['macd'] < (df2["macd"].mean() + df2["macd"].std())):
            return 1
        else:
            return 0
    else:
        return 0
df2['signal'] = df2.apply(signal, axis=1)
```

```python
def analyze_performance(data):
    performance = {
        'negative': sum(data['signal_return'] < 0),
        '1%': sum((data['signal_return'] >= 0) & (data['signal_return'] <= .01)),
        '2%': sum((data['signal_return'] > .01) & (data['signal_return'] <= .02)),
        '3%': sum((data['signal_return'] > .02) & (data['signal_return'] <= .03)),
        '4%': sum((data['signal_return'] > .03) & (data['signal_return'] <= .04)),
        '5%': sum((data['signal_return'] > .04) & (data['signal_return'] <= .05)),
        'Greater than 5%': sum((data['signal_return'] > .05)),
    }
    return pd.DataFrame(data = [performance])

analyze_performance(df2)
```

```python
def analyze_performance(data, timeframe):
    performance = {
        'negative': sum(data[f'signal_return{timeframe}'] < 0),
        '1%': sum((data[f'signal_return{timeframe}'] >= 0) & (data[f'signal_return{timeframe}'] <= .01)),
        '2%': sum((data[f'signal_return{timeframe}'] > .01) & (data[f'signal_return{timeframe}'] <= .02)),
        '3%': sum((data[f'signal_return{timeframe}'] > .02) & (data[f'signal_return{timeframe}'] <= .03)),
        '4%': sum((data[f'signal_return{timeframe}'] > .03) & (data[f'signal_return{timeframe}'] <= .04)),
        '5%': sum((data[f'signal_return{timeframe}'] > .04) & (data[f'signal_return{timeframe}'] <= .05)),
        'Greater than 5%': sum((data[f'signal_return{timeframe}'] > .05)),
    }
    return pd.DataFrame(data = [performance])
```

```python
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans, DBSCAN

df = pd.read_csv('stocks.csv')
df.head()
```

```python
df.dtypes
```

```python
df.duplicated().sum()
df.isna().sum()
```

```python
import matplotlib.pyplot as plt

df.hist(bins=50, figsize=(20, 20))

for ax in plt.gcf().axes:
    ax.set_xlim(df.quantile(0.01, numeric_only=True), df.quantile(0.99, numeric_only=True))

plt.show()
```

```python
df_clipped = df.clip(lower=df.quantile(0.01, numeric_only=True), upper=df.quantile(0.99, numeric_only=True), axis=1)
df_clipped.hist(bins=50, figsize=(20, 20))
plt.show()
```

```python
df.describe()
```

```python
from sklearn.preprocessing import StandardScaler
for i in df.columns:
    if (df[i].dtypes == 'int64') or (df[i].dtypes == 'float64'):
        df[i] = df[i].fillna(df[i].median())
        
df_clipped = df.clip(lower=df.quantile(0.01, numeric_only=True), upper=df.quantile(0.99, numeric_only=True), axis=1)
x = df_clipped.select_dtypes(include=['number'])
# x = df.select_dtypes(include='number')
x_scaled = StandardScaler().fit_transform(x)

clust = KMeans(n_clusters=3, random_state=2).fit(x_scaled)
df['labels'] = clust.labels_
```

```python
df.describe(percentiles=[.01, .99])
```

```python
df['labels'] = clust.labels_
```

```python
import matplotlib.pyplot as plt
coeffs = np.polyfit(df_clipped['Earnings per share diluted, Trailing 12 months'], df_clipped['Price'], deg=2)
poly = np.poly1d(coeffs)

y_fit = poly(df_clipped['Earnings per share diluted, Trailing 12 months'])

plt.scatter(df_clipped['Earnings per share diluted, Trailing 12 months'], df_clipped['Price'], c=df['labels']); 
plt.plot(df_clipped['Earnings per share diluted, Trailing 12 months'], y_fit, c='r')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = df_clipped['Earnings per share diluted, Trailing 12 months']
y = df_clipped['Price']

# Fit
coeffs = np.polyfit(x, y, deg=2)
poly = np.poly1d(coeffs)

# Smooth curve
x_smooth = np.linspace(x.min(), x.max(), 200)
y_smooth = poly(x_smooth)

# Plot
plt.scatter(x, y, c=df['labels'], alpha=0.5)
plt.plot(x_smooth, y_smooth, c='r')
plt.show()
```
